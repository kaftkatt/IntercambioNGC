```python
alpha=2*1e-4
RHO=np.zeros(np.shape(ds.THETA.values))
for i in range(len(ds.T)):
    tempIn=np.full(shape=np.shape(ds.THETA[i].values), fill_value=np.nan)
    tempIn[np.where(mask==False)]=ds.THETA[i].values[np.where(mask==False)]
    RHO[i,:,:,:]=999.8*(1-(tempIn-8)*alpha)
```

```python
i=-1
drho=(RHO[i,1:,:,:]-RHO[i,:-1,:,:])/(dsG.Z.values[1:,None,None]-dsG.Z.values[:-1,None,None])
g=9.81
Nf2=-(g/999.8)*drho
Nf2[Nf2<0]=np.nan
Nf=np.nanmean(np.sqrt(Nf2),axis=1)

zf=(dsG.Z[1:].values+dsG.Z[:-1].values)/2
```

##### Calculate Richardson number $Ri=-g \frac{d\rho}{dz} / \rho (\frac{dv}{dz})^{2} $ using L as the hydraulic depth (crosssectional area/free surface width)

```python
dV=(ds.VVEL[i,1:,umbral1,:].values-ds.VVEL[i,:-1,umbral1,:].values)
dRHO=(RHO[i,1:,umbral1,:]-RHO[i,:-1,umbral1,:])
dz=(dsG.Z.values[1:,None]-dsG.Z.values[:-1,None])
Ri=(-g*np.nanmin(dRHO)/dz[np.argmax(np.abs(dRHO[:,1]))]/999.8)*(np.max(dV)/dz[np.argmax(np.abs(dV[:,1]))]**2)
```
