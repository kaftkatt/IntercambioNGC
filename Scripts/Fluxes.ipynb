```python
from scipy.interpolate import RegularGridInterpolator
```

```python
def getIsoThermVal(levels,val,valTemp):
    caxC=plt.contour(dsG.Y[OB:OBend]/1000,dsG.Z,valTemp,levels=levels)
    
    for i in caxC.get_paths()[:]:
            v = i.vertices
            x = v[:, 0]
            y = v[:, 1]

    interp = RegularGridInterpolator((dsG.Z.values,dsG.Y[OB:OBend].values/1000), val)
    
    ValIsoTherm=interp((y,x))
    
    return ValIsoTherm,x

def fluxes(value, levels):
    valAv=np.sum(np.ma.masked_array(np.mean(value[:,:,OB:OBend,OBx:OBendx],axis=0),mask=mask[:,OB:OBend,OBx:OBendx]),axis=2)
    valW=np.ma.masked_array(np.mean(value[:,:,OB:OBend,OBx],axis=0),mask=mask[:,OB:OBend,OBx])
    valE=np.ma.masked_array(np.mean(value[:,:,OB:OBend,OBendx-1],axis=0),mask=mask[:,OB:OBend,OBendx-1])
    
    valTempAv=np.mean(np.ma.masked_array(np.mean(dsdyn.THETA[:,:,OB:OBend,OBx:OBendx].values,axis=0),mask=mask[:,OB:OBend,OBx:OBendx]),axis=2)
    valTempW=np.ma.masked_array(np.mean(dsdyn.THETA[:,:,OB:OBend,OBx].values,axis=0),mask=mask[:,OB:OBend,OBx])
    valTempE=np.ma.masked_array(np.mean(dsdyn.THETA[:,:,OB:OBend,OBendx-1].values,axis=0),mask=mask[:,OB:OBend,OBendx-1])
    
    IsoAv,xAv=getIsoThermVal(levels,valAv,valTempAv)
    IsoW,xW=getIsoThermVal(levels,valW,valTempW)
    IsoE,xE=getIsoThermVal(levels,valE,valTempE)

    return IsoAv,IsoW,IsoE,xAv,xW,xE
```

```python
Cp = 3.994e03 # Heat capacity in MITgcm (J/kg/K)
rho0 = 1026.35 # density kg/m^3
```

```python
Area=np.ma.masked_array(dsG.HFacC.values*dsG.rA.values[None,:,:],mask=mask)
value=(ds.ADVr_TH.values*Cp*rho0)/Area
valueD=(ds.DFrI_TH.values*Cp*rho0)/Area
```

```python

Av,W,E,xAv,xW,xE=fluxes(value, [8.75])

AvD,WD,ED,xAvD,xWD,xED=fluxes(valueD, [8.75])
```

```python
def BottomConc(vel, hfac, nz,nt, ny,nx,loc):
    '''vel: Velocity field (nt,nz,ny)
       hfac: fraction of open cell at center (nz,ny)
       
       
       RETURNS:
       
       BotVel = velocity near bottom (nt,ny)'''
    
    BotVel = np.empty((nt,ny,nx))
    BottomInd = np.argmax(hfac[::-1,:,5]>0.0,axis=0) # start looking for first no-land cell from the bottom up.
    BottomInd = (np.ones(np.shape(BottomInd))*nz - BottomInd)-loc # Get index of unreversed z axis
    
    for j in range(nt):
        for i in range(ny): 
            for k in range(nx): 
                BotVel[j,i,k] = vel[j,int(BottomInd[i]),i,k] 
                
    
    return (BotVel,BottomInd.astype(int))
```
